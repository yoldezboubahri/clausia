```python
import json

NOISE = [
    re.compile(r"^EN$"),
    re.compile(r"^Official Journal of the European Union$"),
    re.compile(r"^L \d+/\d+$"),
    re.compile(r"^\d{1,2}\.\d{1,2}\.\d{4}$"),
]

def clean_lines(text):
    text = text.replace("\xa0", " ").replace("\xad\n", "").replace("\xad", "")
    out = []
    for l in text.split("\n"):
        l = l.strip()
        if not l or any(p.match(l) for p in NOISE):
            continue
        out.append(l)
    return out

def pack_paragraphs(lines, max_chars=3500):
    """Coupe un long article aux numéros de paragraphe ('1.', '2.' seuls sur une ligne)."""
    groups, cur = [], []
    for l in lines:
        if re.fullmatch(r"\d+\.", l) and cur:
            groups.append(cur)
            cur = []
        cur.append(l)
    if cur:
        groups.append(cur)
    parts, buf = [], ""
    for g in groups:
        t = " ".join(g)
        if buf and len(buf) + len(t) > max_chars:
            parts.append(buf)
            buf = t
        else:
            buf = (buf + " " + t).strip()
    if buf:
        parts.append(buf)
    return parts

sl = clean_lines(solv)

cand = [(i, int(l.split()[1])) for i, l in enumerate(sl) if re.fullmatch(r"Article \d+", l)]
headings, expected = [], 1
for i, n in cand:
    if n == expected:
        headings.append((i, n))
        expected += 1

# La directive finit avant les annexes
last = headings[-1][0]
end_all = next((i for i in range(last, len(sl)) if sl[i] == "ANNEX I"), len(sl))

solv_chunks = []
for k, (start, n) in enumerate(headings):
    end = headings[k + 1][0] if k + 1 < len(headings) else end_all
    title = sl[start + 1]
    body = sl[start + 2 : end]
    parts = pack_paragraphs(body)
    for j, p in enumerate(parts, 1):
        solv_chunks.append({
            "id": f"SII-Art{n}" + (f"-p{j}" if len(parts) > 1 else ""),
            "source": "Directive 2009/138/EC (Solvency II)",
            "article": n,
            "title": title,
            "text": p,
        })

print("Chunks directive :", len(solv_chunks))
print([c["id"] for c in solv_chunks[:8]])
```

```python
il = clean_lines(ifrs)
il = il[next(i for i, l in enumerate(il) if l == "OBJECTIVE"):]

PARA = re.compile(r"^((?:B|C)?\d+[A-Z]?)\.$")

ifrs_chunks, section, cur_id, cur_txt = [], "OBJECTIVE", None, []

def flush():
    if cur_id and cur_txt:
        ifrs_chunks.append({
            "id": f"IFRS17-{cur_id}",
            "source": "IFRS 17 Insurance Contracts (Regulation (EU) 2021/2036)",
            "paragraph": cur_id,
            "section": section,
            "text": " ".join(cur_txt),
        })

for line in il:
    m = PARA.match(line)
    if m:
        flush()
        cur_id, cur_txt = m.group(1), []
    elif line.isupper() and 3 < len(line) and len(line.split()) <= 8:
        section = line
    else:
        cur_txt.append(line)
flush()

print("Chunks IFRS 17 :", len(ifrs_chunks))
print([c["id"] for c in ifrs_chunks[:10]])
print([c["id"] for c in ifrs_chunks[-10:]])
```

```python
# 1. Où la numérotation repart-elle en arrière ?
seq = [(idx, PARA.match(line).group(1)) for idx, line in enumerate(il) if PARA.match(line)]
print("Numéros trouvés :", len(seq))

prev, shown = 0, 0
for idx, pid in seq:
    if pid.isdigit():
        n = int(pid)
        if n < prev and shown < 12:
            print(f"ligne {idx} : retour de {prev} à {n}")
            shown += 1
        prev = n

# 2. Les paragraphes B existent-ils sous une autre forme ?
b_lines = [(i, l[:70]) for i, l in enumerate(il) if re.match(r"^B\d+", l)]
print("\nLignes commençant par B + chiffre :", len(b_lines))
print(b_lines[:8])

# 3. Titres d'appendices
for i, l in enumerate(il):
    if re.match(r"^(APPENDIX|Appendix) [A-D]", l):
        print(i, l[:80])
```
